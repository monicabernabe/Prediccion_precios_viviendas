###**1.5. Suposiciones iniciales:**


- Precisión de los datos: Se asume que los datos proporcionadosson correctos y completos. Se espera que los valores no contengan errores significativos ni inconsistencias importantes.

- Relación lineal entre características y precio: Se supone que existe una relación entre las características de la vivienda y el precio de venta.

- Distribución de precios homogénea: Se asume que el precio de las viviendas se distribuye de manera relativamente homogénea dentro de las áreas geográficas especificadas (por código postal y coordenadas). Aunque puede haber fluctuaciones, se espera que el análisis de los precios siga ciertas tendencias observables.

- Disponibilidad de datos geográficos: Se asume que las coordenadas de latitud y longitud son suficientemente representativas para analizar la ubicación y su influencia en el precio, sin grandes discrepancias o distorsiones debidas a imprecisiones en la recopilación de datos geográficos.

###**1.6. Medición del rendimiento:**

Para evaluar el rendimiento del modelo en este proyecto, se utilizarán varias métricas de regresión que permitirán medir la precisión y efectividad de las predicciones del precio en función de las características de la vivienda, tales como:

- **RMSE (Root Mean Squared Error):** mide la diferencia promedio entre los valores predichos por el modelo y los valores reales del precio de las viviendas. Un valor más bajo de RMSE indica que el modelo es más preciso en sus predicciones.

- **MSE (Mean Squared Error):**calcula el promedio del cuadrado de la diferencia entre los valores estimados y el valor actual.

- **MAE (Mean Absolute Error):** calcula la diferencia promedio en términos absolutos entre las predicciones y los valores reales. Un valor más bajo de MAE es indicativo de un mejor rendimiento del modelo.

- **R² (Coeficiente de Determinación):** proporciona una medida de la calidad del ajuste del modelo. R² varía entre 0 y 1, donde 1 indica una predicción perfecta. Un valor cercano a 1 una mejor calidad en el ajuste del modelo.

- **Validación cruzada:** Se utilizará una técnica de validación cruzada para asegurarse de que el modelo generaliza bien a datos no vistos, minimizando el sobreajuste y garantizando que los resultados obtenidos sean robustos.

Estas métricas proporcionarán una visión detallada del rendimiento del modelo y permitirán realizar ajustes para optimizar su capacidad predictiva en función de los datos disponibles.

###**5.1. Entrenamiento de modelos:**

El problema para el que se entrenará el modelo de machine learning es un problema de regresión, ya que la variable objetivo toma valores contínuos.

Para abordar este problema, se entrenarán distintos modelos de regresión que son adecuados para este tipo de tarea:

- **LinearRegression**: Modelo lineal simple que establece una relación lineal entre la variable objetivo y las variables independientes. Sirve como modelo base o de referencia.

- **Lasso (Least Absolute Shrinkage and Selection Operator)**: Modelo de regresión lineal que incluye una penalización L1, lo que le permite realizar selección de variables, reduciendo a cero los coeficientes menos relevantes.

- **Stochastic Gradient Descent (SGDRegressor)**: Algoritmo de optimización que entrena modelos lineales de forma iterativa, útil especialmente para datasets grandes. Puede combinarse con penalizaciones L1 o L2.

- **SVR (Support Vector Regression)**: Versión del algoritmo de Support Vector Machines adaptada a regresión. Busca una función que se desvíe lo menos posible de los valores reales dentro de un margen tolerable, siendo robusto frente a outliers.

- **KNeighborsRegressor**: Modelo basado en vecinos más cercanos. Predice el valor de una muestra como la media (u otra métrica) de los valores de las muestras más cercanas en el espacio de características. Es útil cuando la relación entre las variables no es lineal.

- **DecisionTreeRegressor**: Modelo de árbol de decisión que divide los datos en subconjuntos en función de valores de las características, ajustando predicciones locales en cada región. Puede sobreajustar si no se controla su profundidad.

- **RandomForestRegressor**: Ensamble de múltiples árboles de decisión que mejora la generalización del modelo y reduce el riesgo de overfitting. Suele ofrecer buen rendimiento con poca necesidad de ajuste fino.

- **GradientBoostingRegressor**: Algoritmo de boosting que entrena modelos secuenciales, donde cada nuevo modelo corrige los errores del anterior. Generalmente ofrece alto rendimiento en problemas de regresión complejos.

Para evaluar y comparar el rendimiento de estos modelos, se analizarán diversas métricas, como:

- **RMSE (Root Mean Squared Error):** mide la diferencia promedio entre los valores predichos por el modelo y los valores reales del precio de las viviendas. Un valor más bajo de RMSE indica que el modelo es más preciso en sus predicciones.

- **MSE (Mean Squared Error):**calcula el promedio del cuadrado de la diferencia entre los valores estimados y el valor actual.

- **MAE (Mean Absolute Error):** calcula la diferencia promedio en términos absolutos entre las predicciones y los valores reales. Un valor más bajo de MAE es indicativo de un mejor rendimiento del modelo.

- **R² (Coeficiente de Determinación):** proporciona una medida de la calidad del ajuste del modelo. R² varía entre 0 y 1, donde 1 indica una predicción perfecta. Un valor cercano a 1 una mejor calidad en el ajuste del modelo.

- **Validación cruzada:** Se utilizará una técnica de validación cruzada para asegurarse de que el modelo generaliza bien a datos no vistos, minimizando el sobreajuste y garantizando que los resultados obtenidos sean robustos.

El análisis de estas métricas permitirá seleccionar el modelo con mejor rendimiento para la tarea de predicción de los precios de las viviendas.
